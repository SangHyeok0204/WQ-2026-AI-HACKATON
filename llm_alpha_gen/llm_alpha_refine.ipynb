# Expression Sanity checker
데이터필드 정보를 이용해서, LLM이 만든 expression이 실제 브레인에서 돌아갈 수 있는지를 체크합니다.

```python

import llm_functions as llm
from parser import *
import ace_lib as ace
import numpy as np
import pandas as pd
import os 
import json
import asyncio



# 🔧 REGION / UNIVERSE GLOBALIZED
datafields = llm.import_json(f'datafield/1/{REGION}/{UNIVERSE}/1_{REGION}_{UNIVERSE}_total.json')
operators = llm.import_json('operators_list.json')

# Brain에서 group argument로 사용되는 특수 식별자
# 파서가 datafield로 분류하지만 실제로는 group 키워드
GROUP_IDENTIFIERS = {'industry', 'subindustry', 'sector', 'market', 'country'}

def return_type(node, operators, datafields):
    if node.node_type == 'operator':
        return operators[node.value]['output']

    elif node.node_type == 'number':
        return "NUMBER"

    elif node.node_type == 'datafield':
        if node.value.lower() == 'nan':
            return "MATRIX"
        # group_rank/group_scale 등에서 사용되는 group 식별자 처리
        if node.value.lower() in GROUP_IDENTIFIERS:
            return "GROUP"
        return datafields[node.value]['type']

    else:
        return "SPECIAL_ARGUMENT"


def check_input(operator_inputs, children_types, _debug = False):
    if _debug: print(operator_inputs)
    while len(operator_inputs) != 0:

        operator_input = operator_inputs.pop(0)
        if _debug: print(f"operator input: {operator_input}, remain inputs: {len(operator_inputs)}" )

        if operator_input == '_SPECIAL_ARGUMENT':
            if len(children_types) == 0:
                pass
            elif children_types[0] == '_SPECIAL_ARGUMENT':
                if _debug: print(operator_input)
                if _debug: print(children_types.pop(0))
            else:
                pass

        else:
            if len(children_types) == 0:
                return False

            children_type = children_types.pop(0)
            if operator_input == "MATRIX" and children_type == "NUMBER":
                pass
            elif operator_input != children_type:
                if _debug: print(operator_input ,children_type, operator_input == children_type)
                return False

    return True


def sanity_checker(exp, _debug=False):

    try:
        exp_tree = tree_node(exp)

        for node in [node for node in exp_tree.collect_all_nodes() if node.node_type == "operator"]:

            if check_input(eval(operators[node.value]['input']), [return_type(x,operators, datafields) for x in node.children]):
                if _debug: print(check_input(eval(operators[node.value]['input']), [return_type(x) for x in node.children]))
            else:
                return False
        if return_type(exp_tree,operators, datafields)=="MATRIX":
            return True

        else:
            return False
    except Exception as e:
        print(f"err: {e}")
        return False
```

```python

#s = ace.start_session()

# 🔧 REGION / UNIVERSE GLOBALIZED - 글로벌 변수 사용
region = REGION
universe = UNIVERSE

datafields = llm.import_json(f'datafield/1/{region}/{universe}/1_{region}_{universe}_total.json')
operators = llm.import_json('operators_list.json')

# ============================================================
# almost_alphas.txt 로드 (아깝게 통과 못한 알파들)
# ============================================================
import re as _re

def load_almost_alphas(filepath='almost_alphas.txt'):
    """almost_alphas.txt를 파싱하여 expression 리스트를 반환."""
    expressions = []
    with open(filepath, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            # "signal = add(...)" 같은 assignment prefix 제거
            m = _re.match(r'^([a-zA-Z_]\w*)\s*=\s*(.+)$', line)
            if m and '(' not in m.group(1):
                line = m.group(2).strip()
            expressions.append(line)
    return expressions

almost_alphas = load_almost_alphas('almost_alphas.txt')

# idea.md 로드
with open('idea.md', 'r', encoding='utf-8') as f:
    idea_content = f.read()

print(f"[INFO] Loaded {len(almost_alphas)} almost-alphas from almost_alphas.txt")
print(f"[INFO] Loaded idea.md ({len(idea_content)} chars)")
print(f"[INFO] First 5 almost-alphas:")
for i, expr in enumerate(almost_alphas[:5]):
    print(f"  [{i}] {expr[:100]}")
```


## 유틸리티 함수
* `sanity_checker()`로 타입 검증 (seed 구조 제약 없음)
* 시뮬레이션 결과 추출, 선별 (sharpe > 0.5, fail_count ≤ 4), 중복 제거
* `final_alphas.txt`로 정렬된 결과 저장


```python

# ============================================================
# Combined-Alpha Pipeline: Utility Functions
# ============================================================
import os
import re
import json
import numpy as np


# ============================================================
# Alpha Selection Criteria for final_alphas.txt
# ============================================================
# SELECTION RULES:
#   Include alpha if ALL of:
#     (1) sharpe > 0.5
#     (2) fail_count <= 4
#
#   Discard if:
#     - sharpe <= 0.5
#     - fail_count >= 5
#     - simulation did not complete (record is None)
# ============================================================

def should_save_alpha(record):
    """
    Check if an alpha record meets the selection criteria for saving.
    """
    if record is None:
        return False
    sharpe = record.get('sharpe', 0)
    if sharpe is None:
        sharpe = 0
    fail_count = record.get('fail_count', 99)
    return sharpe > 0.5 and fail_count <= 4


# ------------------------------------------------------------
# Extract simulation record from Brain API raw response
# ------------------------------------------------------------
def extract_sim_record(sim_result):
    """
    Extract structured dict from a single Brain API simulation result.
    Returns None if the result is invalid or simulation did not complete.
    """
    if not isinstance(sim_result, dict) or 'id' not in sim_result:
        return None
    
    is_data = sim_result.get('is', {})
    if not is_data:
        return None
    
    checks = is_data.get('checks', [])
    
    failed = [c.get('name', '') for c in checks if c.get('result') == 'FAIL']
    warnings = [c.get('name', '') for c in checks
                if c.get('result') not in ('PASS', 'FAIL') and c.get('result')]
    
    for source in (is_data, sim_result):
        extra_w = source.get('warnings')
        if isinstance(extra_w, list):
            warnings.extend(str(w) for w in extra_w)
    
    settings = sim_result.get('settings', {})
    
    return {
        'alpha_id': sim_result['id'],
        'region': settings.get('region', ''),
        'universe': settings.get('universe', ''),
        'sharpe': is_data.get('sharpe', 0),
        'fitness': is_data.get('fitness', 0),
        'turnover': is_data.get('turnover', 0),
        'expression': sim_result.get('regular', {}).get('code', ''),
        'fail_count': len(failed),
        'failed_checks': failed,
        'warnings': warnings,
    }


# ------------------------------------------------------------
# Format one alpha entry for final_alphas.txt
# ------------------------------------------------------------
def format_alpha_entry(idx, record):
    """Format one alpha record into the exact block format."""
    fail_count = record.get('fail_count', 0)
    failed = record.get('failed_checks', [])
    warnings = record.get('warnings', [])
    
    label = "PASS" if fail_count == 0 else "FAIL"
    
    lines = []
    lines.append('-' * 48)
    lines.append(f"--- #{idx} | FAIL: {fail_count} ({label}) ---")
    lines.append(f"ID: {record.get('alpha_id', 'N/A')}")
    lines.append(f"Region: {record.get('region', '')}, Universe: {record.get('universe', '')}")
    lines.append(f"Sharpe: {record.get('sharpe', 0)}, Fitness: {record.get('fitness', 0)}, Turnover: {record.get('turnover', 0)}")
    lines.append(f"Expression: {record.get('expression', '')}")
    
    status_line = f"Status: {label}"
    if warnings:
        status_line += f" (WARNINGS: {', '.join(warnings)})"
    if failed:
        status_line += f" (FAIL_REASONS: {', '.join(failed)})"
    lines.append(status_line)
    
    lines.append('-' * 48)
    
    return '\n'.join(lines)


# ------------------------------------------------------------
# Load existing final_alphas.txt records
# ------------------------------------------------------------
def load_existing_records(filepath='final_alphas.txt'):
    """
    Parse existing final_alphas.txt back into record dicts.
    """
    if not os.path.exists(filepath):
        return []
    
    records = []
    current = None
    
    with open(filepath, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.rstrip('\n')
            
            if line.startswith('=' * 10) or line.startswith('-' * 10):
                continue
            
            m = re.match(r'^--- #\d+ \| FAIL: (\d+)', line)
            if m:
                if current and current.get('alpha_id'):
                    records.append(current)
                current = {
                    'fail_count': int(m.group(1)),
                    'failed_checks': [],
                    'warnings': [],
                    'alpha_id': '', 'region': '', 'universe': '',
                    'sharpe': 0, 'fitness': 0, 'turnover': 0,
                    'expression': '',
                }
                continue
            
            if current is None:
                continue
            
            if line.startswith('ID: '):
                current['alpha_id'] = line[4:].strip()
            
            elif line.startswith('Region: '):
                parts = line.split(', ')
                current['region'] = parts[0].replace('Region: ', '').strip()
                if len(parts) > 1:
                    current['universe'] = parts[1].replace('Universe: ', '').strip()
            
            elif line.startswith('Sharpe: '):
                for part in line.split(', '):
                    kv = part.split(': ', 1)
                    if len(kv) == 2:
                        k = kv[0].strip().lower()
                        try:
                            current[k] = float(kv[1].strip())
                        except (ValueError, TypeError):
                            current[k] = 0
            
            elif line.startswith('Expression: '):
                current['expression'] = line[12:].strip()
            
            elif line.startswith('Status: '):
                status_str = line[8:]
                warn_match = re.search(r'\(WARNINGS: ([^)]+)\)', status_str)
                if warn_match:
                    current['warnings'] = [w.strip() for w in warn_match.group(1).split(',')]
                fail_match = re.search(r'\(FAIL_REASONS: ([^)]+)\)', status_str)
                if fail_match:
                    current['failed_checks'] = [f.strip() for f in fail_match.group(1).split(',')]
            
            elif line.startswith('Failed: '):
                current['failed_checks'] = [x.strip() for x in line[8:].split(',')]
    
    if current and current.get('alpha_id'):
        records.append(current)
    
    return records


# ------------------------------------------------------------
# Write sorted, filtered, deduplicated final_alphas.txt
# ------------------------------------------------------------
def write_final_txt(records, region, universe, filepath='final_alphas.txt'):
    """
    Filter, deduplicate, sort, and write entries to final_alphas.txt.
    """
    filtered = [r for r in records if should_save_alpha(r)]
    
    filtered.sort(key=lambda r: (r.get('fail_count', 99), -(_safe_float(r.get('sharpe')))))
    
    seen_expr = set()
    unique_records = []
    for r in filtered:
        expr = r.get('expression', '').strip()
        if expr and expr not in seen_expr:
            seen_expr.add(expr)
            unique_records.append(r)
    
    unique_records.sort(key=lambda r: (r.get('fail_count', 99), -(_safe_float(r.get('sharpe')))))
    
    lines = []
    lines.append('=' * 80)
    lines.append(f"COMBINED ALPHA LIST (Sorted: fail_count ASC, sharpe DESC)")
    lines.append(f"Region: {region}, Universe: {universe}")
    lines.append(f"Selection: sharpe > 0.5 AND fail_count <= 4")
    lines.append(f"Total: {len(unique_records)} alphas")
    lines.append('=' * 80)
    lines.append('')
    
    for idx, record in enumerate(unique_records, 1):
        lines.append(format_alpha_entry(idx, record))
        lines.append('')
    
    pass_count = sum(1 for r in unique_records if r.get('fail_count', 0) == 0)
    fail_count = sum(1 for r in unique_records if r.get('fail_count', 0) >= 1)
    warn_count = sum(1 for r in unique_records if r.get('warnings'))
    best_sharpe = max((_safe_float(r.get('sharpe')) for r in unique_records), default=0)
    
    lines.append('=' * 80)
    lines.append('Summary:')
    lines.append(f"- Total: {len(unique_records)} alphas")
    lines.append(f"- PASS (fail_count=0): {pass_count}")
    lines.append(f"- FAIL (fail_count>=1): {fail_count}")
    lines.append(f"- With warnings: {warn_count}")
    lines.append(f"- Best Sharpe: {best_sharpe}")
    lines.append('=' * 80)
    
    with open(filepath, 'w', encoding='utf-8') as f:
        f.write('\n'.join(lines))
    
    print(f"[WRITTEN] {filepath} ({len(unique_records)} alphas | PASS: {pass_count}, FAIL: {fail_count}, warnings: {warn_count})")
    return unique_records


def _safe_float(val):
    """Convert a value to float safely, returning 0.0 for None or invalid."""
    if val is None:
        return 0.0
    try:
        return float(val)
    except (ValueError, TypeError):
        return 0.0



# ------------------------------------------------------------
# Submit-ready alphas: 0-fail AND no concentrated weight
# → submit_ready.txt 로 별도 저장
# ------------------------------------------------------------
SUBMIT_READY_FILE = 'submit_ready.txt'

def is_submit_ready(record):
    """
    fail_count == 0 AND warnings에 concentrated weight 없는 알파.
    """
    if record is None:
        return False
    if record.get('fail_count', 99) \!= 0:
        return False
    for w in record.get('warnings', []):
        if 'concentrated' in w.lower():
            return False
    return True


def write_submit_ready_txt(records, region, universe, filepath=SUBMIT_READY_FILE):
    """
    0-fail + no concentrated weight 알파만 필터링하여 별도 파일로 저장.
    Sharpe 내림차순 정렬, expression 기준 중복 제거.
    """
    seen_expr = set()
    unique = []
    for r in records:
        expr = r.get('expression', '').strip()
        if expr and expr not in seen_expr:
            seen_expr.add(expr)
            unique.append(r)

    ready = [r for r in unique if is_submit_ready(r)]
    ready.sort(key=lambda r: -(_safe_float(r.get('sharpe'))))

    lines = []
    lines.append('=' * 80)
    lines.append(f"SUBMIT-READY ALPHAS (0-fail, no concentrated weight)")
    lines.append(f"Region: {region}, Universe: {universe}")
    lines.append(f"Total: {len(ready)} alphas")
    lines.append('=' * 80)
    lines.append('')

    for idx, record in enumerate(ready, 1):
        lines.append(format_alpha_entry(idx, record))
        lines.append('')

    best_sharpe = max((_safe_float(r.get('sharpe')) for r in ready), default=0)

    lines.append('=' * 80)
    lines.append('Summary:')
    lines.append(f"- Total: {len(ready)} submit-ready alphas")
    lines.append(f"- Best Sharpe: {best_sharpe}")
    lines.append('=' * 80)

    with open(filepath, 'w', encoding='utf-8') as f:
        f.write('
'.join(lines))

    print(f"[WRITTEN] {filepath} ({len(ready)} submit-ready alphas | Best Sharpe: {best_sharpe})")
    return ready
```


# 시뮬레이션 돌리기
* `almost_alphas.txt` + `idea.md` 패턴으로 LLM이 조합 알파 생성
* 생성 비율: A(10%) 직접조합 / B(80%) idea.md 패턴 / C(10%) 변형+조합
* 배치당 80개 생성 → 8개씩 미니배치 시뮬레이션
* 결과는 `final_alphas.txt`에 저장


```python

# ============================================================
# Combined-Alpha Generation Pipeline (Resume 지원)
# - almost_alphas.txt의 알파들을 idea.md 패턴으로 조합/변형
# - sanity_checker()로 구문 검증 후 시뮬레이션
# - 결과를 final_alphas.txt에 저장 (incremental merge)
#
# GENERATION RATIO: A(10%) 직접조합 / B(80%) idea.md 패턴 / C(10%) 변형+조합
# SELECTION: sharpe > 0.5 AND fail_count <= 4
# SORTING:   fail_count ASC, sharpe DESC
# DEDUP:     by expression
# ============================================================

import os
import json
import re
import llm_functions as llm
import ace_lib as ace

# 🔧 CONFIG
ALPHA_NUM = 80           # LLM에게 요청할 알파 개수 per batch
NUM_BATCHES = 10          # 총 LLM 배치 수 (필요시 증가)
MINI_BATCH_SIZE = 8      # 시뮬레이션 미니배치 크기
OUTPUT_FILE = 'final_alphas.txt'
GEN_DIR = './gen_json'
GEN_PREFIX = 'combined'  # gen_json/combined_0.json, combined_1.json, ...


# ------------------------------------------------------------
# Resume 지원 함수들
# ------------------------------------------------------------

def get_existing_json_count(prefix, gen_dir="./gen_json"):
    if not os.path.exists(gen_dir):
        os.makedirs(gen_dir, exist_ok=True)
        return 0
    pattern = re.compile(rf"^{re.escape(prefix)}_(\d+)\.json$")
    count = 0
    for filename in os.listdir(gen_dir):
        if pattern.match(filename):
            count += 1
    return count


def get_next_json_index(prefix, gen_dir="./gen_json"):
    if not os.path.exists(gen_dir):
        os.makedirs(gen_dir, exist_ok=True)
        return 0
    pattern = re.compile(rf"^{re.escape(prefix)}_(\d+)\.json$")
    max_k = -1
    for filename in os.listdir(gen_dir):
        match = pattern.match(filename)
        if match:
            k = int(match.group(1))
            if k > max_k:
                max_k = k
    return max_k + 1


def save_json_safe(data, filepath):
    if os.path.exists(filepath):
        print(f"[WARN] File already exists, skipping: {filepath}")
        return False
    with open(filepath, 'w', encoding='utf-8') as f:
        json.dump(data, f, ensure_ascii=False, indent=2)
    return True


# ------------------------------------------------------------
# operators_list.json → compact string for LLM prompt
# ------------------------------------------------------------

def format_operators_compact(operators_dict):
    ops_by_category = {}
    for op_name, op_info in operators_dict.items():
        cat = op_info.get('category', 'Other')
        if cat not in ops_by_category:
            ops_by_category[cat] = []
        ops_by_category[cat].append(op_info.get('definition', op_name))

    lines = []
    for cat, defs in ops_by_category.items():
        lines.append(f"[{cat}] " + " | ".join(defs))
    return "\n".join(lines)


# ------------------------------------------------------------
# Combined-Alpha LLM prompt & generation
# ------------------------------------------------------------

def generate_combined_expressions(s, alpha_region, alpha_universe,
                                   almost_alphas_list, idea_md_content,
                                   model='gpt-5-mini-2025-08-07',
                                   alpha_num=80):
    """
    almost_alphas + idea.md 패턴을 기반으로 조합 알파를 생성.
    A(10%) 직접조합 / B(80%) idea.md 패턴 / C(10%) 변형+조합
    """
    # operators compact string
    operators_dict = llm.import_json('operators_list.json')
    operators_compact = format_operators_compact(operators_dict)

    # almost_alphas를 번호 매긴 문자열로
    numbered_alphas = "\n".join(
        f"[{i}] {expr}" for i, expr in enumerate(almost_alphas_list)
    )

    # A/B/C 개수 계산
    a_count = max(1, round(alpha_num * 0.10))
    c_count = max(1, round(alpha_num * 0.10))
    b_count = alpha_num - a_count - c_count

    answer_form = '''{
"results":[
    {
        "idea":...,
        "description":...,
        "implementation":...,
        "confidence_level":...
    }
]
}'''

    prompt = f"""
<MISSION>
You have a list of "almost-alpha" expressions below. These are quantitative trading alpha
formulas that NEARLY passed WorldQuant Brain's simulation checks but fell short on
Sharpe ratio, fitness, turnover, or correlation criteria.

Your task: Using the ALPHA_IDEAS patterns, generate {alpha_num} NEW alpha expressions by
combining, transforming, and enhancing these almost-alphas.

Each result must have 4 parts:
- "idea": which type (A/B/C) and core concept
- "description": which almost-alphas (by index [N]) inspired this, what transformation applied
- "implementation": the FASTEXPR expression (MUST be syntactically valid)
- "confidence_level": 0-1 scale
</MISSION>

<GENERATION_RATIO>
Type A (~{a_count} alphas, ~10%): DIRECT COMBINATION
  Combine 2-3 almost-alphas using basic operators (add, subtract, multiply, divide).
  Example: add(multiply([alpha_5], 0.6), multiply([alpha_12], 0.4))
  Goal: blend complementary signals for diversification.

Type B (~{b_count} alphas, ~80%): IDEA-DRIVEN GENERATION
  Use the ALPHA_IDEAS patterns (Templates A/B/C, weighted combination, ratio signals,
  group_rank/group_scale with industry/subindustry, signed_power) to create new alphas.
  Use the almost-alphas as BUILDING BLOCKS within these templates.
  Example: signed_power(group_rank(min([alpha_8], [alpha_22]), industry), 0.8)
  Goal: apply proven successful structures to promising signals.

Type C (~{c_count} alphas, ~10%): VARIATION + STRUCTURAL COMBINATION
  Apply minor variations to almost-alphas (change window params, swap similar operators),
  then combine using:
  - min(max(X, -1), 1) for signal clipping
  - group_rank(..., industry/subindustry) for group normalization
  - signed_power(..., 0.5~0.8) for signal dampening
  - group_scale(..., subindustry) for group scaling
  Goal: fine-tune and stabilize existing signals.
</GENERATION_RATIO>

<ALMOST_ALPHAS>
{numbered_alphas}
</ALMOST_ALPHAS>

<ALPHA_IDEAS>
{idea_md_content}
</ALPHA_IDEAS>

<OPERATORS_LIST>
{operators_compact}
</OPERATORS_LIST>

<RULES>
1. You may use ANY datafield that appears in the ALMOST_ALPHAS list above.
2. You may also use datafields recommended in ALPHA_IDEAS (momentum ranks, risk fields, etc.).
3. For group_rank/group_scale, use `industry` or `subindustry` as the group argument.
4. VECTOR type fields (vec_avg, vec_sum, etc. are already wrapped in almost-alphas) must
   always be wrapped with vec_* operators before use.
5. signed_power exponent: 0.3 ~ 0.9.
6. Time-series windows: 5 ~ 252 (diversify across different lookback periods).
7. All expressions must be valid FASTEXPR syntax — no Python assignment (=), no semicolons.
8. Reference almost-alphas broadly — use signals from ALL parts of the list, not just the first few.
9. Each generated alpha must be UNIQUE — no duplicates, no copy of existing almost-alphas.
10. For Type B: directly incorporate the structural patterns from ALPHA_IDEAS (group_rank + min/max,
    weighted add with rank(), divide with +1 denominator, ts_decay_linear, etc.)
</RULES>

<ANSWER_FORMAT>
You must answer in this form:
{answer_form}
</ANSWER_FORMAT>
""".strip()

    result_json = llm.call_llm_stream(prompt, answer_form, model)
    return result_json


# ============================================================
# 메인 루프 (Combined-Alpha Pipeline)
# ============================================================

os.makedirs(GEN_DIR, exist_ok=True)

region = REGION
universe = UNIVERSE

print(f"[CONFIG] Region: {region}, Universe: {universe}")
print(f"[CONFIG] Alpha per batch: {ALPHA_NUM}, Batches: {NUM_BATCHES}")
print(f"[CONFIG] Simulation mini-batch: {MINI_BATCH_SIZE}")
print(f"[CONFIG] Selection: sharpe > 0.5 AND fail_count <= 4")
print(f"[CONFIG] Output: {OUTPUT_FILE}")

# 기존 결과 로드 (incremental merge)
all_records = load_existing_records(OUTPUT_FILE)
print(f"[INFO] Loaded {len(all_records)} existing records from {OUTPUT_FILE}")

# 시작 인덱스 결정
start_index = get_next_json_index(GEN_PREFIX, GEN_DIR)
existing_count = get_existing_json_count(GEN_PREFIX, GEN_DIR)

print(f"[INFO] Existing JSON files: {existing_count}, Starting at index: {start_index}")

for batch_idx in range(NUM_BATCHES):
    
    current_index = start_index + batch_idx
    save_path = f"{GEN_DIR}/{GEN_PREFIX}_{current_index}.json"
    
    if os.path.exists(save_path):
        print(f"[SKIP] {save_path} already exists")
        continue
    
    print(f"\n{'='*60}")
    print(f"[BATCH {batch_idx+1}/{NUM_BATCHES}] index={current_index}")
    print(f"{'='*60}")
    
    # 1. LLM Generate
    print("--- Generating combined alphas with LLM ---")
    try:
        response = generate_combined_expressions(
            s, region, universe,
            almost_alphas, idea_content,
            model='gpt-5-mini-2025-08-07',
            alpha_num=ALPHA_NUM
        )
    except Exception as e:
        print(f"[ERROR] LLM generation failed: {e}")
        continue
    
    print("--- Received response ---")
    
    # 2. Parse JSON
    try:
        json_response = llm.cut_first_to_last_brace(response)
        alphas = json.loads(json_response)
    except Exception as e:
        print(f"[ERROR] JSON parsing failed: {e}")
        continue
    
    # 3. Save raw JSON
    if not save_json_safe(alphas, save_path):
        print(f"[WARN] Could not save {save_path}, skipping")
        continue
    print(f"[SAVED] {save_path}")
    
    # 4. Filter with sanity_checker (seed 구조 제약 없이 타입 검증만)
    raw_results = alphas.get("results", [])
    candidates = []
    rejected = 0
    
    for x in raw_results:
        expr = x.get('implementation', '')
        if sanity_checker(expr):
            candidates.append(x)
        else:
            rejected += 1
    
    print(f"[FILTER] {len(candidates)} passed sanity check, {rejected} rejected out of {len(raw_results)}")
    
    if len(candidates) == 0:
        print(f"[WARN] No valid alphas after sanity check")
        continue
    
    # 5. Build alpha list for simulation
    alpha_list = [
        ace.generate_alpha(
            regular=x['implementation'],
            region=region,
            universe=universe,
            neutralization="INDUSTRY"
        )
        for x in candidates
    ]
    
    tags_list = [
        ['combined_alpha', f"confidence_{x.get('confidence_level', 0)}"]
        for x in candidates
    ]
    descs_list = [x.get('description', '') for x in candidates]
    
    len_alpha_list = len(alpha_list)
    print(f"[INFO] Simulating {len_alpha_list} combined alphas ({MINI_BATCH_SIZE} per mini-batch)")
    
    # 6. Simulate in mini-batches
    for rep in range(0, len_alpha_list, MINI_BATCH_SIZE):
        
        # Session timeout check
        if ace.check_session_timeout(s) < 500:
            print("[INFO] Session timeout approaching, refreshing...")
            s = ace.start_session()
        
        batch_end = min(rep + MINI_BATCH_SIZE, len_alpha_list)
        
        try:
            print(f"--- Simulating mini-batch {rep} to {batch_end} ---")
            
            sim_results = list(
                ace.multi_simulate_alphas_map(
                    s,
                    alpha_list[rep:batch_end],
                    tags_list[rep:batch_end],
                    descs_list[rep:batch_end],
                    batch_end - rep
                )
            )
            
            batch_total = 0
            batch_saved = 0
            for result in sim_results:
                record = extract_sim_record(result)
                if record:
                    batch_total += 1
                    all_records.append(record)
                    if should_save_alpha(record):
                        batch_saved += 1
            
            print(f"  [SELECT] {batch_saved}/{batch_total} pass selection criteria")
            
        except Exception as e:
            print(f"[ERROR] Simulation failed: {e}")
            continue
    
    # 7. Write sorted, filtered, deduplicated final_alphas.txt after each batch
    write_final_txt(all_records, region, universe, OUTPUT_FILE)
    write_submit_ready_txt(all_records, region, universe)

print(f"\n{'='*60}")
print("[DONE] Combined-Alpha pipeline finished")
print(f"{'='*60}")
```


# 기존 JSON에서 시뮬레이션만 돌리기
* `gen_json/combined_*.json`에 JSON은 생성되었지만 시뮬레이션이 안 돌아간 경우 사용
* 기존 JSON 파일을 읽어서 `sanity_checker()` 필터링 후 시뮬레이션만 수행
* 결과는 `final_alphas.txt`에 저장


```python

# ============================================================
# 기존 JSON에서 시뮬레이션만 돌리기 (Simulate-Only from existing JSON)
# - gen_json/combined_*.json 파일을 읽어서
# - sanity_checker() 필터링 후
# - 시뮬레이션만 수행하고 결과를 final_alphas.txt에 저장
# ============================================================

import os
import re
import json
import ace_lib as ace

GEN_DIR = "./gen_json"
OUTPUT_FILE = "final_alphas.txt"
GEN_PREFIX = "combined"
MINI_BATCH_SIZE = 8

region = REGION
universe = UNIVERSE

def get_json_files_for_prefix(prefix, gen_dir="./gen_json"):
    """prefix에 해당하는 모든 JSON 파일 경로를 반환 (인덱스 순 정렬)."""
    if not os.path.exists(gen_dir):
        return []
    pattern = re.compile(rf"^{re.escape(prefix)}_(\d+)\.json$")
    files = []
    for filename in os.listdir(gen_dir):
        match = pattern.match(filename)
        if match:
            files.append((int(match.group(1)), os.path.join(gen_dir, filename)))
    files.sort(key=lambda x: x[0])
    return files

print(f"[CONFIG] Region: {region}, Universe: {universe}")
print(f"[CONFIG] Selection: sharpe > 0.5 AND fail_count <= 4")
print(f"[CONFIG] Mode: Simulate-only from existing JSON files (prefix: {GEN_PREFIX})")

json_files = get_json_files_for_prefix(GEN_PREFIX, GEN_DIR)

print(f"\n{'='*60}")
print(f"[INFO] JSON files found: {len(json_files)}")
print(f"{'='*60}")

if len(json_files) == 0:
    print(f"[SKIP] No JSON files with prefix '{GEN_PREFIX}' found in {GEN_DIR}")
else:
    # 기존 txt 레코드 로드 (이미 시뮬레이션된 expression 스킵용)
    all_records = load_existing_records(OUTPUT_FILE)
    simulated_exprs = set(r.get('expression', '').strip() for r in all_records if r.get('expression'))
    print(f"[INFO] Loaded {len(all_records)} existing records from {OUTPUT_FILE}")
    print(f"[INFO] {len(simulated_exprs)} unique expressions already simulated")
    
    for file_idx, json_path in json_files:
        
        print(f"\n--- Processing {os.path.basename(json_path)} ---")
        
        # 1. JSON 로드
        try:
            with open(json_path, 'r', encoding='utf-8') as f:
                alphas = json.load(f)
        except Exception as e:
            print(f"[ERROR] Failed to load {json_path}: {e}")
            continue
        
        # 2. sanity_checker 필터링
        raw_results = alphas.get("results", [])
        candidates = []
        rejected = 0
        skipped_dup = 0
        
        for x in raw_results:
            expr = x.get('implementation', '').strip()
            
            # 이미 시뮬레이션된 expression 스킵
            if expr in simulated_exprs:
                skipped_dup += 1
                continue
            
            if sanity_checker(expr):
                candidates.append(x)
            else:
                rejected += 1
        
        print(f"[FILTER] {len(candidates)} to simulate, {rejected} invalid, {skipped_dup} already simulated (total: {len(raw_results)})")
        
        if len(candidates) == 0:
            print(f"[SKIP] No new candidates to simulate")
            continue
        
        # 3. 시뮬레이션용 alpha list 생성
        alpha_list = [
            ace.generate_alpha(
                regular=x['implementation'],
                region=region,
                universe=universe,
                neutralization="INDUSTRY"
            )
            for x in candidates
        ]
        
        tags_list = [
            ['combined_alpha', f"confidence_{x.get('confidence_level', 0)}"]
            for x in candidates
        ]
        descs_list = [x.get('description', '') for x in candidates]
        
        len_alpha_list = len(alpha_list)
        print(f"[INFO] Simulating {len_alpha_list} alphas")
        
        # 4. 미니배치 시뮬레이션
        for rep in range(0, len_alpha_list, MINI_BATCH_SIZE):
            
            if ace.check_session_timeout(s) < 500:
                print("[INFO] Session timeout approaching, refreshing...")
                s = ace.start_session()
            
            batch_end = min(rep + MINI_BATCH_SIZE, len_alpha_list)
            
            try:
                print(f"--- Simulating mini-batch {rep} to {batch_end} ---")
                
                sim_results = list(
                    ace.multi_simulate_alphas_map(
                        s,
                        alpha_list[rep:batch_end],
                        tags_list[rep:batch_end],
                        descs_list[rep:batch_end],
                        batch_end - rep
                    )
                )
                
                batch_total = 0
                batch_saved = 0
                for result in sim_results:
                    record = extract_sim_record(result)
                    if record:
                        batch_total += 1
                        all_records.append(record)
                        simulated_exprs.add(record.get('expression', '').strip())
                        if should_save_alpha(record):
                            batch_saved += 1
                
                print(f"  [SELECT] {batch_saved}/{batch_total} pass selection criteria")
                
            except Exception as e:
                print(f"[ERROR] Simulation failed: {e}")
                continue
        
        # 5. 각 JSON 파일 처리 후 txt 갱신
        write_final_txt(all_records, region, universe, OUTPUT_FILE)
        write_submit_ready_txt(all_records, region, universe)
    
    print(f"\n{'='*60}")
    print("[DONE] Simulate-only pipeline finished")
    print(f"{'='*60}")
```
