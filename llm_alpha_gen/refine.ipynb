```python
# ============================================================
# 0. IMPORTS
# ============================================================
import os
import re
import ast
import sys
import time
import random
import json
from dotenv import load_dotenv
from openai import OpenAI

# ============================================================
# 1. CONFIG
# ============================================================
MODEL_NAME = "gpt-4.0"


GENERATION_SIZE = 50        # 🔥 생성 = 50
MAX_GENERATIONS = 100
TARGET_SHARPE = 2.5

REGION = "EUR"
UNIVERSE = "TOP2500"
TEST_PERIOD = "P2Y"
DELAY = 1
NEUTRALIZATION = "INDUSTRY"

PARENT_POOL_SIZE = 30
LOOP_SLEEP = 1

# ============================================================
# 2. OPENAI CLIENT
# ============================================================
# 현재 노트북 파일 위치 기준으로 .env.local 로드
load_dotenv(os.path.join(os.getcwd(), ".env.local"), override=True)

client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
print("API KEY LOADED:", os.getenv("OPENAI_API_KEY")[:10] if os.getenv("OPENAI_API_KEY") else "NOT FOUND")

# ============================================================
# 2.5 LOAD OPERATORS FROM operators_list.json
# ============================================================
def load_operators_list(filepath="operators_list.json"):
    """
    operators_list.json에서 operator 정보를 로드하여
    LLM 프롬프트용 문자열로 변환
    """
    with open(filepath, 'r', encoding='utf-8') as f:
        operators_dict = json.load(f)
    
    # 카테고리별로 그룹화
    categories = {}
    for name, info in operators_dict.items():
        category = info.get('category', 'Other')
        if category not in categories:
            categories[category] = []
        
        definition = info.get('definition', name)
        description = info.get('description', '')
        categories[category].append(f"{definition}: {description}")
    
    # 프롬프트용 문자열 생성
    operators_text = ""
    for category, ops in sorted(categories.items()):
        operators_text += f"\n{category} Operators:\n"
        for op in ops:
            operators_text += f"  {op}\n"
    
    return operators_text

# operators_list.json 로드
try:
    OPERATORS_TEXT = load_operators_list("operators_list.json")
    print(f"[INFO] Loaded operators from operators_list.json")
except FileNotFoundError:
    print("[WARN] operators_list.json not found, using hardcoded operators")
    OPERATORS_TEXT = """
Arithmetic Operators:
  add(x, y): Element-wise addition of two signals.
  subtract(x, y): Element-wise subtraction x−y.
  multiply(x, y): Element-wise multiplication.
  divide(x, y): Element-wise ratio.
  abs(x): Absolute value.
  sign(x): Directional indicator.
  log(x): Natural logarithm.
  sqrt(x): Square root transformation.
  inverse(x): Reciprocal 1/x.
  min(x, y): Element-wise minimum.
  max(x, y): Element-wise maximum.
  power(x, y): Raises x to power y.
  signed_power(x, y): Power transform preserving sign.
  reverse(x): Negates the signal −x.

Time Series Operators:
  ts_backfill(x, d): Fills missing values within d days.
  ts_mean(x, d): Rolling mean over d days.
  ts_delta(x, d): Difference from d days ago.
  ts_zscore(x, d): Rolling z-score.
  ts_rank(x, d): Time-series rank over d days.
  ts_std_dev(x, d): Rolling standard deviation.
  ts_sum(x, d): Rolling sum.
  ts_corr(x, y, d): Rolling correlation.
  ts_covariance(x, y, d): Rolling covariance.
  ts_decay_linear(x, d): Linearly decayed weighted average.
  ts_ir(x, d): Information ratio.
  ts_returns(x, d): Percentage change over d days.
  ts_scale(x, d): Scales to [0,1] using rolling min/max.
  ts_av_diff(x, d): Deviation from rolling mean.
  ts_arg_max(x, d): Days since maximum.
  ts_arg_min(x, d): Days since minimum.
  ts_delay(x, d): Value lagged by d days.

Cross-Sectional Operators:
  rank(x): Cross-sectional rank.
  zscore(x): Cross-sectional standardization.
  winsorize(x): Clips extreme values.
  scale(x): Scales to target book size.
  normalize(x): Mean-centers the signal.
  quantile(x): Maps ranks to distribution.
"""

# ============================================================
# 3. UTILS
# ============================================================
def parse_expression_list(text):
    m = re.search(r"expression_list\s*=\s*(\[[\s\S]*?\])", text)
    if not m:
        return None
    try:
        return ast.literal_eval(m.group(1))
    except:
        return None


def tournament_select(pool):
    if len(pool) < 2:
        return random.choice(pool)
    a, b = random.sample(pool, 2)
    return a if a["sharpe"] > b["sharpe"] else b


def clip_parent_pool(pool, max_size):
    pool = sorted(pool, key=lambda x: x["sharpe"], reverse=True)
    return pool[:max_size]


# ============================================================
# 4. PROMPT (operators_list.json 사용)
# ============================================================
def build_prompt(base_expr, parent_a=None, parent_b=None):
    parents = ""
    if parent_a:
        parents += f"\nPARENT A:\n{parent_a}\n"
    if parent_b:
        parents += f"\nPARENT B:\n{parent_b}\n"

    return f"""
You are a WorldQuant BRAIN quantitative researcher.

BASE ALPHA:
{base_expr}
{parents}

TASK:
- Generate {GENERATION_SIZE} DIFFERENT alpha expressions
- Each must start with "signal =" and end with ";"
- Use ONLY Fast Expression syntax
- Time-series windows must be >= 63
- Allow deeper nesting as generations progress
- Avoid trivial sign flips only

ALLOWED OPERATORS (Fast Expression only):
{OPERATORS_TEXT}

Return ONLY:

expression_list = [
  "signal = ...;",
  ...
]

INPUT EXAMPLE:

BASE ALPHA:
signal = rank(signed_power(ts_mean(ts_backfill(mdl25_vrv421_81v, 63), 126), 2));


OUTPUT EXAMPLE:

expression_list = [
  "signal = ts_rank(ts_delta(ts_backfill(mdl25_vrv421_81v, 63), 20), 252);",
  "signal = ts_rank(add(ts_delta(ts_backfill(mdl25_vrv421_81v, 63), 20),(ts_mean(ts_backfill(mdl25_vrv421_81v, 63), 20)), 252);"
]
"""


# ============================================================
# 5. BASE ALPHA INPUT
# ============================================================
BASE_EXPR = input(
    "signal = rank(signed_power(ts_mean(ts_backfill(mdl25_vrv421_81v, 63), 126), 2));"
).strip()

# ============================================================
# 6. EVOLUTION LOOP
# ============================================================
parent_pool = []
best_sharpe = -999
best_expr = None

for gen in range(1, MAX_GENERATIONS + 1):
    print(f"\n================ Generation {gen} =================")

    parent_a = tournament_select(parent_pool)["expr"] if parent_pool and random.random() < 0.7 else BASE_EXPR
    parent_b = tournament_select(parent_pool)["expr"] if parent_pool and random.random() < 0.4 else None

    prompt = build_prompt(BASE_EXPR, parent_a, parent_b)

    try:
        response = client.chat.completions.create(
            model=MODEL_NAME,
            messages=[
                {"role": "system", "content": "You are a quantitative finance expert."},
                {"role": "user", "content": prompt}
            ],
        )

        text = response.choices[0].message.content
    except Exception as e:
        print("❌ openai call failed:", e)
        continue

    expression_list = parse_expression_list(text)
    if not expression_list or len(expression_list) < 5:
        print("❌ expression_list parsing failed")
        continue

    expression_list = expression_list[:GENERATION_SIZE]

    print(f"Generated {len(expression_list)} alphas")

    # ========================================================
    # 7. CREATE + SIMULATE ALL 50
    # ========================================================
    try:
        alpha_list = [
            ace.generate_alpha(
                x,
                region=REGION,
                universe=UNIVERSE,
                test_period=TEST_PERIOD,
                delay=DELAY,
                neutralization=NEUTRALIZATION
            )
            for x in expression_list
        ]
    except Exception as e:
        print("❌ Alpha creation failed:", e)
        continue

    try:
        result = ace.simulate_alpha_list_multi(s, alpha_list)
    except Exception as e:
        print("❌ Simulation failed:", e)
        continue

    # ========================================================
    # 8. EVALUATION
    # ========================================================
    gen_best = (-999, None)

    for i, r in enumerate(result):
        if r is None:
            continue
        
        if r.get("alpha_id") is None:
          continue

        if r.get("is_stats") is None:
          print(f"  [WARN] Result {i}: is_stats is None")
          continue
        sharpe = float(r["is_stats"]["sharpe"].values[0])
        turnover = float(r["is_stats"]["turnover"].values[0])
        expr = expression_list[i]

        print(f"[{i:02d}] Sharpe={sharpe:.3f}, Turnover={turnover:.3f}")

        parent_pool.append({
            "expr": expr,
            "sharpe": sharpe,
            "turnover": turnover,
            "alpha_id": r["alpha_id"]
        })

        if sharpe > gen_best[0]:
            gen_best = (sharpe, expr)

        if sharpe > best_sharpe:
            best_sharpe = sharpe
            best_expr = expr

        if sharpe >= TARGET_SHARPE:
            print("\n🔥 TARGET SHARPE ACHIEVED!")
            print("Best Alpha:", expr)
            print("Sharpe:", sharpe)
            sys.exit(0)

    parent_pool = clip_parent_pool(parent_pool, PARENT_POOL_SIZE)

    print(f"\n✅ Generation {gen} done")
    print(f" - Best this gen: Sharpe={gen_best[0]:.3f}")
    print(f" - Best overall:  Sharpe={best_sharpe:.3f}")
    print(f" - Parent pool size: {len(parent_pool)}")

    time.sleep(LOOP_SLEEP)

print("\n❌ MAX GENERATIONS reached")
print("Best Alpha:", best_expr)
print("Best Sharpe:", best_sharpe)
```
